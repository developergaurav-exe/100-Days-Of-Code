### Scraping Live website

````python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://news.ycombinator.com/")
yc_webpage = response.text

soup = BeautifulSoup(yc_webpage, "html.parser")

soup.find(name="span", class_= "titleline").get_text()
````

````python
articles = soup.find_all(name="span", class_= "titleline")

article_links = []
for link in articles:

    article_links.append(link.find("a").get("href"))
    # print(link.find("a").get("href"))

print(article_links)
````

````python
upvotes = soup.find_all(name="span", class_= "score")
l = []
for number in upvotes:
    l.append(int(number.get_text().split()[0]))

highest_upvote_index = l.index(max(l))

print(articles[highest_upvote_index].find("a").get_text()) # this will show article who has highest upvotes in first page
print(articles[highest_upvote_index].find("a").get("href")) # this will show article link of highest upvotes in first page
````

````python
article_link = soup.find(name="span", class_= "titleline").find("a").get("href")
article_score = soup.find(name="span", class_= "score")
````

````python
import requests
from bs4 import BeautifulSoup

response = requests.get("https://web.archive.org/web/20200518073855/https://www.empireonline.com/movies/features/best-movies-2/") # web archive URL
movie_webpage = response.text

soup = BeautifulSoup(movie_webpage, "html.parser")

titles = soup.find_all(name="h3", class_="title")

watchlist = []
for movies in titles:
    watchlist.append(movies.get_text()) #.split(')')[1])

print(watchlist)

with open("watchlist.txt", 'w') as file:
	for line in watchlist:
          try:
              file.write(line + '\n')
            
          except UnicodeEncodeError: # 59) E.T. â\x80\x93 The Extra Terrestrial issue here so skipped this moview
              pass 
````
